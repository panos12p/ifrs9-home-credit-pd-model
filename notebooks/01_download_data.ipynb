This notebook requires Kaggle API credentials and acceptance of the competition rules.
A 401 response means the current environment is not authorized to download the data.

```python
import os
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW_DIR = PROJECT_ROOT / "data" / "raw"

RAW_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DIR)
print("Kaggle token exists:", os.getenv("KAGGLE_API_TOKEN") is not None)
```

```python
import kagglehub
from kagglehub.exceptions import KaggleApiHTTPError

try:
    path = kagglehub.competition_download("home-credit-default-risk")
    print(path)
except KaggleApiHTTPError as exc:
    if exc.response is not None and exc.response.status_code == 401:
        print("Kaggle download skipped: HTTP 401 (credentials or competition authorization required).")
    else:
        raise
```
